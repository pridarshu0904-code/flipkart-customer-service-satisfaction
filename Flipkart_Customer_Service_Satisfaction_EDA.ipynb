## 3. ***Data Wrangling***

### Data Wrangling Code

````python
df_clean = df.copy()

date_columns = [
    'order_date_time',
    'Issue_reported at',
    'issue_responded',
    'Survey_response_Date'
]

df_clean[date_columns].head()
````

````python
df_clean['order_date_time'] = pd.to_datetime(
    df_clean['order_date_time'],
    format='%d/%m/%Y %H:%M',
    errors='coerce'
)

df_clean['Issue_reported at'] = pd.to_datetime(
    df_clean['Issue_reported at'],
    format='%d/%m/%Y %H:%M',
    errors='coerce'
)

df_clean['issue_responded'] = pd.to_datetime(
    df_clean['issue_responded'],
    format='%d/%m/%Y %H:%M',
    errors='coerce'
)

df_clean['Survey_response_Date'] = pd.to_datetime(
    df_clean['Survey_response_Date'],
    format='%d-%b-%y',
    errors='coerce'
)
````

### What all manipulations have you done and insights you found?

Data wrangling was performed by creating a separate copy of the original dataset to preserve the raw data. The date-related columns, including order date and time, issue reported time, issue responded time, and survey response date, were converted from text format to appropriate datetime format. Invalid or unparseable values were handled as missing values during conversion. These transformations make the date fields suitable for further analysis such as time-based trends and response-time analysis while keeping the original dataset unchanged.
