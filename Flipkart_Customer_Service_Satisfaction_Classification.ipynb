## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

**Hypothetical Statement 1:** Customer satisfaction (CSAT) differs across the three support channels: Inbound, Outcall, and Email.

**Hypothetical Statement 2:** Customer satisfaction (CSAT) differs across the different agent tenure groups.

**Hypothetical Statement 3:** There is a significant monotonic relationship between Item Price and CSAT Score.

### **Hypothetical Statement - 1**

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis (H₀):** The mean CSAT scores are equal across Inbound, Outcall, and Email support channels.

**Alternate Hypothesis (H₁):** At least one support channel has a different mean CSAT score from the others.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import f_oneway

inbound_csat = df_clean.loc[
    df_clean['channel_name'] == 'Inbound',
    'CSAT Score'
]

outcall_csat = df_clean.loc[
    df_clean['channel_name'] == 'Outcall',
    'CSAT Score'
]

email_csat = df_clean.loc[
    df_clean['channel_name'] == 'Email',
    'CSAT Score'
]

f_statistic, p_value = f_oneway(
    inbound_csat,
    outcall_csat,
    email_csat
)

print("F-statistic:", f_statistic)
print("P-value:", p_value)
````

##### Which statistical test have you done to obtain P-Value?

One-way ANOVA (Analysis of Variance).

##### Why did you choose the specific statistical test?

One-way ANOVA was chosen because we are comparing the mean CSAT scores across three independent support-channel groups: Inbound, Outcall, and Email. This test determines whether there is a statistically significant difference in the mean CSAT scores among these groups.

### **Hypothetical Statement - 2**

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis (H₀):** The mean CSAT scores are equal across all agent tenure groups.

**Alternate Hypothesis (H₁):** At least one agent tenure group has a different mean CSAT score from the others.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import f_oneway

tenure_groups = [
    group['CSAT Score'].values
    for _, group in df_clean.groupby('Tenure Bucket')
]

f_statistic, p_value = f_oneway(*tenure_groups)

print("F-statistic:", f_statistic)
print("P-value:", p_value)
````

##### Which statistical test have you done to obtain P-Value?

One-way ANOVA (Analysis of Variance).

##### Why did you choose the specific statistical test?

One-way ANOVA was chosen because we are comparing the mean CSAT scores across multiple independent agent tenure groups. This test determines whether there is a statistically significant difference in the mean CSAT scores among the different tenure groups.

### **Hypothetical Statement - 3**

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis (H₀):** There is no significant monotonic relationship between Item Price and CSAT Score.

**Alternate Hypothesis (H₁):** There is a significant monotonic relationship between Item Price and CSAT Score.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import spearmanr

correlation_data = df_clean[
    ['Item_price', 'CSAT Score']
].dropna()

spearman_statistic, p_value = spearmanr(
    correlation_data['Item_price'],
    correlation_data['CSAT Score']
)

print("Spearman correlation coefficient:", spearman_statistic)
print("P-value:", p_value)
````

##### Which statistical test have you done to obtain P-Value?

Spearman's rank correlation test.

##### Why did you choose the specific statistical test?

Spearman's rank correlation test was chosen because we are examining the relationship between Item Price and CSAT Score, and the hypothesis specifically concerns a monotonic relationship. This non-parametric test measures the strength and direction of a monotonic association between the two variables without requiring them to have a normal distribution.

### 2. Handling Outliers

````python
# Handling outliers using IQR-based capping

outlier_cols = [
    'Item_price',
    'connected_handling_time'
]

for col in outlier_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_clean[col] = df_clean[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )

print("Outlier treatment completed using IQR-based capping.")
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

IQR-based capping (Winsorization) was used to treat outliers in Item Price and Connected Handling Time. The IQR method identifies observations outside the lower and upper bounds based on the first and third quartiles. Instead of removing these observations, capping was used to limit extreme values to the calculated IQR boundaries while retaining the customer interaction records. This approach was selected because Item Price and Connected Handling Time contain valid numerical observations with extreme values, and removing them could result in loss of useful information. CSAT Score was not treated for outliers because its values from 1 to 5 represent valid customer satisfaction classes rather than numerical measurement errors.

#### 1. Expand Contraction

````python
# Expand common English contractions in Customer Remarks

contractions = {
    "can't": "cannot",
    "won't": "will not",
    "n't": " not",
    "'re": " are",
    "'s": " is",
    "'m": " am",
    "'ll": " will",
    "'d": " would",
    "'ve": " have",
    "'t": " not",
    "'em": " them"
}

def expand_contractions(text):
    text = str(text)
    for contraction, expansion in contractions.items():
        text = text.replace(contraction, expansion)
        text = text.replace(contraction.upper(), expansion)
    return text

df_clean['Customer Remarks'] = df_clean['Customer Remarks'].apply(
    expand_contractions
)

print("Contraction expansion completed.")
````

````python
from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBClassifier

# Hyperparameter search space for XGBoost
xgb_param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.05, 0.1, 0.2],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0],
    'min_child_weight': [1, 3, 5]
}

# Base XGBoost model
xgb_base = XGBClassifier(
    objective='multi:softmax',
    num_class=5,
    eval_metric='mlogloss',
    random_state=42,
    n_jobs=-1
)

# RandomizedSearchCV with 3-fold cross-validation
xgb_random_search = RandomizedSearchCV(
    estimator=xgb_base,
    param_distributions=xgb_param_grid,
    n_iter=5,
    scoring='f1_macro',
    cv=3,
    random_state=42,
    n_jobs=-1
)

# Fit the tuning process
xgb_random_search.fit(X_train_transformed, y_train_xgb)

# Select the best model
best_model_3 = xgb_random_search.best_estimator_

# Predict using the tuned model
y_pred_3_tuned = best_model_3.predict(X_test_transformed)

# Convert predictions back to CSAT scores 1-5
y_pred_3_tuned = y_pred_3_tuned + 1

print("Best Parameters:")
print(xgb_random_search.best_params_)

print("\nBest Cross-Validation Macro F1:")
print(round(xgb_random_search.best_score_, 4))

print("\nTuned Model 3 prediction completed.")
````
